### Objetivos de la clase
- 1 - Activar Conocimientos
- 2 - Conocer Distancia
- 3 - Naive Bayes (Algoritmos de Clasificación)

```python
from scipy.spatial import distance_matrix
```

```python
movies = df.columns.values.tolist()[1:]
```

```python
movies
```

```python
# p = exponente 1 manhatan
dmanhattan = distance_matrix(df[movies],df[movies], p = 1) 
dEuclidiana = distance_matrix(df[movies],df[movies], p = 2) 
dEnesima = distance_matrix(df[movies],df[movies], p = 10) 
```

```python
dmanhattan
```

```python
def dm_to_df(dd,col_name):
    return pd.DataFrame(dd,index = col_name, columns=col_name)
```

```python
dm_to_df(dmanhattan,df['user_id'])
```

```python
dm_to_df(dEuclidiana,df['user_id'])
```

```python
dm_to_df(dEnesima,df['user_id'])
```

### El método Euclidiano es el más utilizado

P(clase/dato) = (P(dato/clase) * P(clase)) / P(dato)
