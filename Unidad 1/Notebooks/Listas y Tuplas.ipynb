## Tuplas

```python
tupla = (4) # Las tuplas son INMUTABLES 
tupla
```

```python
# Generamos una tupla con varios valores
tupla2 = (2,3,4,5,6)
lista_vacia = [] # Declaramos una lista vacía
for x in tupla2:
    # lista_nueva.append(x)
    lista_vacia = lista_vacia + [x]
lista_vacia
```

```python
# Tupla a partir de una lista
lista_nueva = [lista1, lista2 , tupla2]
lista_nueva
```

```python
#  
tupla_nueva = (lista2, lista3, tupla2)
tupla_nueva
```

```python
# Podemos modificar la tupla si dentro de ella existe una lista, pero no podemos modifica directamente
lista2.append('Valor nuevo')
```

### Diccionarios

```python
persona = {
    "rut" : "20281676-2",
    "nombre" : "Jorge",
    "apellido" : "Quintui",
    "edad" : 21
}

for x in persona.values():
    print(x)
```

```python
for x, y in persona.items(): 
    print(x,y)
```

```python
for x in persona: 
    print(x)
```

```python
persona[3] = 'hola' 
```

```python
persona
```

```python
del persona[3]
persona
```
