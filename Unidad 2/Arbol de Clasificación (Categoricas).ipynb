```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import datasets
import warnings
```

```python
dataset = datasets.load_breast_cancer()
df = pd.DataFrame(np.c_[dataset['data'],dataset['target']],columns = np.append(dataset['feature_names'],['target']))
```

```python
# preparamos los datos
X = df.drop(["target"], axis = 1)
y = df.target
```

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# Separamos datos de entrenamiento y datos de prueba
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size = 0.2, random_state = 35)

# RANDOM STATE REORDENA LOS DATOS DE FORMA RANDOM 35

# Creamos el modelo
modelo = DecisionTreeClassifier(random_state=35)

# Entremamos el modelo
modelo.fit(X_train,y_train)

# Valido el modelo
y_pred = modelo.predict(X_test)

# Evaluamos el resultado del modelo
print('Precisión del modelo :',modelo.score(X_test,y_test))
print('Accuracy del modelo :',accuracy_score(y_test, y_pred))
```

### Es un modelo casi perfecto

```python
from sklearn.tree import plot_tree

fig, ax = plt.subplots(figsize =(30,10))

plot = plot_tree(
    decision_tree=modelo,
    feature_names=df.drop(columns=["target"]).columns,
    class_names="target",
    filled = True,
    impurity=False,
    fontsize=12,
    precision=2,
    ax = ax
)
```
