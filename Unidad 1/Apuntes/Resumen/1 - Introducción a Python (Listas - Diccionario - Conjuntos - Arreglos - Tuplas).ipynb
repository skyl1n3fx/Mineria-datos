```python
 # Operadores de asignación
x = 15
print("x = ", x)
x += 2
print("x += 2 :", x)
x -= 6
print("x -= 6 :", x)
x *= 3
print("x *= 3 :", x)
x **= 2
print("x **= 2 :", x)
x //= 2
print("x //= 2 :", x)
x /= 3
print("x /= 3 :", x)
x %= 2
print("x %= 2 :", x)
```

```python
# Operadores a nivel de bits
print(5 & 3) # Representa el operador AND
print(5 | 3) # Representa el operador OR
print(5 ^ 3) # Representa el operador XOR
print(5 << 2) # n << x : representa el movimiento de x bits a la izquierda
print(64 >> 2) # n >> x : representa el movimiento de x bits a la derecha
```

### 1.1.3 Explicación de resultados
#### Lo primero es buscar la representación en binario de ambos números 5 (101) y 3 (011)
- *Usando la tabla de verdad del AND 101 AND 011 = 001 que, en el sistema decimal, es el 1*
- *Usando la tabla de verdad del OR 101 AND 011 = 111 que, en el sistema decimal, es el 7*
- *Usando la tabla de verdad del XOR 101 AND 011 = 110 que, en el sistema decimal, es el 6*
- *101 << 2 implica mover 2 bits a la izquierda y queda 10100 que, en el sistema decimal, es 20 = (16+0+4+0+0)*
- *1000000 >> 2 implica mover 2 bits a la derecha y queda 10000 que, en el sistema decimal, es 16*

## 2.2 Diccionarios
- Conjunto de elementos no ordenados
- Contienen clave - valor


```python
# Creación de un diccionario, tiene una estructura similar a un JSON
persona = {
    "rut" : "3232534-3",
    "nombre" : "Jorge",
    "apellido" : "Quintui",
    "edad" : 21
}

# Imprimo los valors del diccionario
for x in persona.values():
    print(x)
```

```python
# Imprimimos clave y valor del diccionario
for x, y in persona.items(): 
    print(x,y)
```

```python
# Imprimimos las claves del diccionario
for x in persona.keys(): 
    print(x)
```

```python
# Agregamos un valor al diccionario, especificando clave y valor
persona['correo'] = 'jorge@gmail.com'
persona
```

```python
# Eliminamos la clave de un diccionario
del persona['correo']
persona
```

```python
# Diccionario concatenado
diccionario2 = {'Docente' : {1:'luis'} , 'Estudiante' : {2:'Jorge'}} 
diccionario2
```

```python
# Diccionario concatenado 2
diccionario3 = {'Docente' : {1 : {'Luis' : 'Diaz' }}}
diccionario3
```

# 3 Numpy & Arreglos (Arrays)
#### Numpy es una extensión de python, que permite trabajar con funciones matemáticas de alto nivel para operar com vectores o matrices.

## Manejo de Arrays
- Transformar un lista en un array (función array)
- Función arange
- Función ones
- Función linspace
- Función eye

```python
import numpy as np
```

```python
# Transformar lista en arreglo
# Arreglo Unidimensional
lista_array = [1,2,3,4,5,6,7,8]
array = np.array(lista_array)
array
```

```python
# Verificamos el tipo
print(type(lista_array[2]))
```

```python
# Cambiamos el tipo de dato a float32
arr2 = np.array(lista_array,dtype="float32")
arr2
```

```python
# Trasformar una lista con distintos valores en un arreglo
temp = [1,'dos',True]
arrayTemp = np.array(temp)
# <U11 único string combinado
arrayTemp
```

```python
# Reshape transforma mi arreglo (Redimensiona), pero la cantidad de elementos debe ser par
# Matriz bidimensional
arr3 = array.reshape(2,4)
arr3
```

```python
# Matriz tridimensional
exp2 = [1,2,3,4,5,6,7,8]
arrexp0 = np.array(exp2)
arrepx = arrexp0.reshape(2,2,2)
arrepx
```

```python
# Función arange
# Permite crear un arreglo de x cantidad de elementos comenzando por el 0
arreglo = np.arange(5)
arreglo
```

```python
# puede recibir más argumentos, determino el inicio
arreglo = np.arange(1,5)
arreglo
```

```python
# Establezco los saltos entre cada valor
arreglo = np.arange(0,10,3)
arreglo
```

```python
# Función zeros
# Permite generar un arreglo lleno de ceros
arreglo = np.zeros([2,3]) # Indico filas y columnas
arreglo
```

```python
# Función ones
# Permite generar un arreglo lleno de unos
arreglo = np.ones([2,3])
arreglo
```

```python
# Función linspace
# Funcion (valor inicial, valor final, cantidad de valores)
# Permite generar un arreglo formado por n números equiespaciados
arreglo = np.linspace(5,8,20)
arreglo
```

```python
# Función eye
# Devuelve una matriz de 2 dimensiones con números unos en la diagonal, y ceros en las demás posiciones
arreglo = np.eye(5)
arreglo
```

```python
# Ejemplo de creación de matriz 333
# 1. Debemos multiplicar los valores 3 x 3 x 3 = 27
# Luego usamos la función arange con dicha cantidad de elementos
matriz333 = np.arange(27)
matriz333
```

```python
# 2. Debemos utilizar la función reshape para darle la estructura tridimensional al array
arreglo333 = matriz333.reshape(3,3,3)
arreglo333
```

```python
# 3. Si queremos indicar un valor dentro de la matriz debemos localizar, Grupo, Fila y Columna
# Intentamos localizar el número 13
# Grupo [0 , 1 , 2]
# Fila [0 , 1 , 2]
# Columna [0 , 1 , 2]
# El número 13 se localiza en el grupo 1, en la fila 1 y en la columna 1
print(arreglo333[1][1][1])
```

```python
# Otro ejemplo
# Matriz 4, 5 ,5 
# Calculamos 4 * 5 * 5 = 100
matriz455 = np.arange(100)
matriz455
```

```python
# Le damos la estructura a la matriz
arreglo455 = matriz455.reshape(4,5,5)
arreglo455
```

```python
# busquemos el número 97
# Grupo - Fila - Columna
print(arreglo455[3][4][2])
```

```python
# Seleccionar elementos dentro de un array mediante una condición
arregloRandom = np.random.randint(1,100,21)
arregloRandom
```

```python
# Elementos que son menores a 10
arregloRandom[arregloRandom<10]
```

```python
# Buscar números pares dentro del arreglo
arregloRandom[arregloRandom % 2 == 0]
```

```python
# Presentar valores que sean mayores a 10 y menores de 50
arregloRandom[(arregloRandom > 10) & (arregloRandom < 50)]
```

##### Función MAX y MIN en arreglos

```python
# Nos devuelve el valor máximo
arregloRandom.max()
```

```python
#  Nos devuelve el valor minímo
arregloRandom.min()
```

```python
# Nos devuelve la posición (index) dónde está el elemento con el valor MAX
arregloRandom.argmax()
```

```python
# Nos devuelve la posición (index) dónde está el elemento con el valor MIN
arregloRandom.argmin()
```
