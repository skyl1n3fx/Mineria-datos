## Naive Bayes

```python
import pandas as pd
import numpy as np
from sklearn import datasets
```

```python
dataset = datasets.load_breast_cancer()

df = pd.DataFrame(
    np.c_[dataset['data'],
    dataset['target']],
    columns = np.append(dataset['feature_names'],['target'])
)
```

** Revisar la distribucion respecto a la variable que usaremos para la ** Clasificacion ** **

```python
# Tamaño de mi variable objetivo
print(df.groupby('target').size()) 
```

```python
import matplotlib.pyplot as plt
```

```python
clases = np.array([df[df.target == 0.0].shape[0],df[df.target == 1.0].shape[0]])
```

```python
clases
```

```python
labels = [str(round(x*1.0/clases.sum()*100.0,2))+'%' for x in clases]
```

```python
labels[0] = 'Benignas \n' + labels[0]
labels[1] = 'Malignas \n' + labels[1]
```

```python
plt.pie(clases, labels=labels)
plt.title('Porcentaje de tumores')
plt.show()
```
