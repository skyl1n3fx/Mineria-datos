```python
from numpy import array
from numpy import argmax
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import MinMaxScaler
```

```python
# defina data de ejemplo
data = ['frio','frio','templado','frio','calor','calor','templado','frio','calor','templado']
values = array(data)
print('Valores ' , values )
```

```python
# Label encoder
# Se transforman las variables categoricas a variables numericas 
label_encoder = LabelEncoder()
integer_encoder = label_encoder.fit_transform(values)
print('Label encoder integer: ' , integer_encoder)
```

```python
# OneHotEncoder
onehot_encoder = OneHotEncoder(sparse = False)
integer_encoded = integer_encoder.reshape(len(integer_encoder),1)
onehot_encoder = onehot_encoder.fit_transform(integer_encoded)
print('one-hot-encoder : \n', onehot_encoder)
```
