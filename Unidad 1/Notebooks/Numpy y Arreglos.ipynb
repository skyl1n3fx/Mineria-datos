## Numpy y Arreglos (arrays)

```python
# Conjunto de datos de un mismo tipo
# Importo la librería numpy 
import numpy as np
```

```python
lista = [1,2,3,4,5,6,7,8]
lista
```

```python
# Se convierte la lista en un arreglo
arr1 = np.array(lista)
arr1
```

```python
print(type(lista[2]))
```

```python
# Cambiamos el tipo de dato a float32
arr2 = np.array(lista,dtype="float32")
arr2
```

```python
# Trasformar una lista con distintos valores en un arreglo
temp = [1,'dos',True]
arrayTemp = np.array(temp)
# <U11 único string combinado
arrayTemp
```

```python
arr1
```

```python
arr3 = arr1.reshape(2,4)
arr3
```

```python
# Para usar reshape se debe tener un arreglo par
exp = [1,2,3,4,5,6,7]
arrexp0 = np.array(exp)
arrexp = arrexp0.reshape(2,4)
arrexp
```

```python
arr4 = arr1.reshape(4,2)
print(arr4)
```

```python
# Matriz tridimensional
exp2 = [1,2,3,4,5,6,7,8]
arrexp0 = np.array(exp2)
arrepx = arrexp0.reshape(2,2,2)
arrepx
```
