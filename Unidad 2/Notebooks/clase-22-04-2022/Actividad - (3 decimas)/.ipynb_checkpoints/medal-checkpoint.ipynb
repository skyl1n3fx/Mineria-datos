```python
# Importamos las librerías necesarias
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import datasets, linear_model
from sklearn.metrics import mean_squared_error, r2_score
%matplotlib inline
```

```python
# Leemos el dataset
data_url = './dataset/Medals.csv'
df = pd.read_csv(data_url, sep=',',encoding= 'unicode_escape')
df
```

```python
df.describe()
```

```python
def graficoCajaBigotes(feature,labelx, labely,title):
    plt.boxplot(df[feature])
    plt.ylabel(labely)
    plt.xlabel(labelx)
    plt.title(title)
    
def graficoLineas(xvalue,yvalue,title):
    df.plot(kind="scatter",x=xvalue,y=yvalue,title=title)
```

```python
graficoLineas(1,4,"Hola")
```
